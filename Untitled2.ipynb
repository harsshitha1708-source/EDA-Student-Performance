# Exploratory Data Analysis of Student Academic Performance

## 1. Introduction

This project performs Exploratory Data Analysis (EDA) on a dataset containing information about student academic performance.

The dataset contains 395 student records and 33 features covering demographic information, family background, study habits, academic history, social factors, and final academic performance.

The main objective of this analysis is to explore the dataset, identify patterns and trends, calculate statistical summaries, visualize relationships between variables, and examine factors associated with students' final grades.

## 2. Problem Statement

Student academic performance can be influenced by several academic, demographic, social, and family-related factors.

The objective of this project is to explore student performance data and identify patterns, relationships, and important factors associated with students' final grades.

The analysis focuses particularly on the final grade (G3) and examines its relationship with study time, absences, previous failures, parental education, previous grades, internet access, and other student-related factors.

## 3. Objectives

The main objectives of this project are:

1. Understand the structure and characteristics of the dataset.

## 5. Dataset Description

The dataset contains information about 395 students and 33 features.

The variables include:

- Demographic information
- Family information
- Study habits

- School-related information
- Social activities

- Previous grades

The primary target variable considered in this analysis is:

**G3 — Final Grade**

The variables G1 and G2 represent earlier-period grades and are also analyzed to understand their relationship with the final grade.

```python
# Display the key results needed for the final report

print("========== FINAL EDA RESULTS ==========\n")

print("Average Final Grade:", round(df["G3"].mean(), 2))
print("Minimum Final Grade:", df["G3"].min())
print("Maximum Final Grade:", df["G3"].max())

print("\nAverage Study Time by Category:")
print(df.groupby("studytime")["G3"].mean())

print("\nAverage Final Grade by Past Failures:")
print(df.groupby("failures")["G3"].mean())

print("\nAverage Final Grade by Gender:")
print(df.groupby("sex")["G3"].mean())

print("\nAverage Final Grade by Internet Access:")
print(df.groupby("internet")["G3"].mean())

print("\nAverage Final Grade by Higher Education Plans:")
print(df.groupby("higher")["G3"].mean())

print("\n========== CORRELATION WITH G3 ==========")
print(df.select_dtypes(include="number").corr()["G3"].sort_values(ascending=False))
```

```python
# Complete correlation results with Final Grade

correlations = df.select_dtypes(include="number").corr()["G3"].drop("G3")

print("Correlation of each numerical variable with Final Grade (G3):")
print(correlations.sort_values(ascending=False))
```

## 6. Key Findings

The exploratory analysis produced the following findings:

1. The average final grade (G3) of the students is 10.42 out of 20, with grades ranging from 0 to 20.

2. G2 has the strongest positive correlation with the final grade, with a correlation coefficient of 0.905.

3. G1 also shows a strong positive correlation with the final grade, with a correlation coefficient of 0.801.

4. The number of previous failures has a negative correlation with the final grade (-0.360). The average final grade decreases across the groups with increasing numbers of previous failures in this dataset.

5. Mother's education (Medu) has a positive correlation of 0.217 with the final grade, while father's education (Fedu) has a correlation of 0.152.

6. Study time has a relatively small positive correlation of 0.098 with the final grade.

7. Absences have a correlation of 0.034 with the final grade, indicating a very small linear relationship in this dataset.

8. Students with internet access have an average final grade of 10.62, compared with 9.41 for students without internet access.

9. Students who reported wanting higher education have an average final grade of 10.61, compared with 6.80 among those who did not.

10. The analysis shows that student performance is associated with several academic and background variables, although correlation alone does not establish causation.

## 7. Conclusion

This Exploratory Data Analysis examined 395 student records and 33 features related to academic performance, demographics, family background, study habits, and social factors.

The analysis used statistical summaries, bar charts, histograms, box plots, scatter plots, and correlation analysis to identify patterns and relationships in the data.

The results show strong positive relationships between earlier grades (G1 and G2) and the final grade (G3). Previous failures show a negative relationship with final performance. Other variables such as parental education, study time, internet access, and higher-education plans also show differences or associations with final grades in this dataset.

Overall, the analysis demonstrates how Exploratory Data Analysis can be used to understand a dataset, discover patterns, compare groups, and identify relationships between variables.

The findings describe relationships observed in this dataset and should not be interpreted as proof that any particular factor directly causes changes in academic performance.
